### utilisation des tableaux d'octets : bytes

```python
# ---------------- chaines de cartactères ASCII en octet ------------------

message = "bonjour les gens"

b_message = b"bonjour les gens"
# encodage   , pas d'encodage
type(message), type(b_message)
```

```python
# ---------------- interface du type bytes ----------------------
## REM: interfaces similaires (capitalize, upper, replace, ...)

#       bytes       ==         bytes    <=>   str
bytes(message, "utf-8"), message.encode(), b_message.decode()
```

```python
# -------------- caractères non ASCII --------------------------

# le préfixe "b" ne foncitonne pas avec des caractère non ascii
# b_message = b'bonjour à tout le monde'
b_message = bytes('bonjour à tout le monde', "utf-8")
b_message
```

```python
# ---------------------- conversions d'octets --------------------
# un objet bytes est une liste de codes ascii (ints)
for code in b_message:
    print(code)

# on peut créer un objet bytes à partir d'une liste de code ASCII
alpha = [ i for i in range(65, 91) ]
bytes(alpha)
```

```python
# ---------------------------- conversion entre code ascii <-> str ---------------

char_a, char_z   = "a", "z"
print(ord(char_a), ord(char_z))

code_a, code_z = 97, 122
print(chr(code_a), chr(code_z)) 
```

```python
# ---------------------------- conversion entre bytes <-> hexadécimal ------------

print(b_message.hex(), bytes.fromhex("626f6e6a6f757220c3a020746f7574206c65206d6f6e6465"))
```

```python
#  ------------------ MINI - EXO: BRUTE FORCE d'un hash ----------------------------

# signature numérique (fonction de hachage)
from hashlib import md5
from string import ascii_lowercase
# module standard qui contient des itérateurs remarquables
import itertools

# je signe la chaine "salut" avec md5 (32 caractères)
# il n'ya pas de méthode inverse pour retrouver la str
hash = md5(bytes("salut", "utf-8")).hexdigest()
print(hash)

# retrouver salut à partir du hash

## je teste les hash des str de len 1 a -> z
##   //                             2 aa -> zz
##   //                             3 aaa -> zzz
## ....

## itérateur count => 1, 2, 3, ---> inf
cracked = False
for i in itertools.count(1):
    ## je dois générer le produit cartésien (toutes les associations entre i ensembles)
    for tup in itertools.product(ascii_lowercase, repeat=i):
        candidate = "".join(tup)
        if md5(bytes(candidate, "utf-8")).hexdigest() == hash:
            print(f"{candidate} cracked !")
            cracked = True
            break
    if cracked: break





```
